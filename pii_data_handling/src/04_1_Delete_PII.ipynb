```python
%sql

CREATE OR REPLACE TABLE db_academy.pii_data.customers_rows_filtering AS
SELECT * FROM db_academy.pii_data.customers_raw
```

```python
%sql

SELECT DISTINCT country FROM db_academy.pii_data.customers_rows_filtering
```

```python
%sql

CREATE OR REPLACE FUNCTION db_academy.pii_data.country_filter(country STRING)
RETURN 
    CASE WHEN is_account_group_member('manager') THEN TRUE
         WHEN country = 'Canada' THEN TRUE
         ELSE FALSE END;
```

```python
%sql

ALTER TABLE db_academy.pii_data.customers_rows_filtering
SET ROW FILTER db_academy.pii_data.country_filter ON (country)   
```
