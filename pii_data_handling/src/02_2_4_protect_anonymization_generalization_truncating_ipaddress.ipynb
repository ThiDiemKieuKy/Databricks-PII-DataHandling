```python
from pyspark.sql.functions import regexp_replace, col

web_log_trunc_ipaddress = (
    spark.read.table("db_academy.pii_data.web_logs_raw")
    .withColumn("new_ip_adress", regexp_replace(col("ip_address"), r"\d+$", "0"))
    .select("log_id", "user_id", "ip_address", "new_ip_adress", "device", "location", "search_query", "timestamp")
)

display(web_log_trunc_ipaddress.limit(10))
```
