```python
from pyspark.sql.functions import col, floor, current_date, months_between, when, lit

# grouping age defination
customers_bins = (
    spark.read.table("db_academy.pii_data.customers_raw")
    .withColumn("age", floor(months_between(current_date(), col("date_of_birth")) / 12 ))
    .withColumn("age_range", when(col("age") < 18, lit("under 18"))
                            .when(col("age") < 25, lit("18-25"))
                            .when(col("age") < 35, lit("25-35"))
                            .when(col("age") < 45, lit("35-45"))
                            .when(col("age") < 55, lit("45-55"))
                            .when(col("age") < 65, lit("55-65"))
                            .when(col("age") < 75, lit("65-75"))
                            .otherwise(lit("65+")))
    .select (
        "customer_id",
        "full_name",
        "email",
        "phone",
        "date_of_birth",
        "age_range" )        
)
display(customers_bins)
```
