```python
from pyspark.sql.functions import floor, col

employees_rounding = (
    spark.read.table("db_academy.pii_data.employees_raw")
    .withColumn("Salary_Rounding", floor(col("salary") / 1000) * 1000)
    .select("employee_id", "name", "job_title", "department", "age", "salary", "Salary_Rounding", "city")
)

display(employees_rounding.limit(10))
```
