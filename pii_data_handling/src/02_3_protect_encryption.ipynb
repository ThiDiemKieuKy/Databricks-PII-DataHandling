```python
from pyspark.sql.functions import col, aes_encrypt, base64, lit

# The aes_encrypt function requires a binary key of exactly 16, 24, or 32 bytes. The string "MyEncryptionKey" is 15 characters
encrypted_key = "MyEncryptionKey1"
customers_encrypted = (
    spark.read
    .table("db_academy.pii_data.customers_raw")
    .select(
        "customer_id",
        base64(aes_encrypt(col("full_name"), lit(encrypted_key))).alias("full_name"),
        base64(aes_encrypt(col("email"), lit(encrypted_key))).alias("email"),
        base64(aes_encrypt(col("phone").cast("string"), lit(encrypted_key))).alias("phone"),
        base64(aes_encrypt(col("ssn"), lit(encrypted_key))).alias("ssn"),
        base64(aes_encrypt(col("date_of_birth").cast("string"), lit(encrypted_key))).alias("date_of_birth"),
        base64(aes_encrypt(col("address"), lit(encrypted_key))).alias("address"),
        base64(aes_encrypt(col("city"), lit(encrypted_key))).alias("city"),
        "country"
    )
)
display(customers_encrypted.limit(10))
```

```python
from pyspark.sql.functions import aes_decrypt, unbase64, col, lit

customers_descrypted = (
    spark.read
    .table("db_academy.pii_data.customers_encrypted")
    .select(
        "customer_id",
        aes_decrypt(unbase64(col("full_name")), lit(encrypted_key)).cast("string").alias("full_name"),
        aes_decrypt(unbase64(col("email")), lit(encrypted_key)).cast("string").alias("email"),
        aes_decrypt(unbase64(col("phone")), lit(encrypted_key)).cast("string").alias("phone"),
        aes_decrypt(unbase64(col("ssn")), lit(encrypted_key)).cast("string").alias("ssn"),
        aes_decrypt(unbase64(col("date_of_birth")), lit(encrypted_key)).cast("string").alias("date_of_birth"),
        aes_decrypt(unbase64(col("address")), lit(encrypted_key)).cast("string").alias("address"),
        aes_decrypt(unbase64(col("city")), lit(encrypted_key)).cast("string").alias("city")
    )
)
display(customers_descrypted.limit(10))
```
